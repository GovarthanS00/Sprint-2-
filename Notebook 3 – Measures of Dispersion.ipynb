## 1. Range
### Definition

Range is the difference between the largest value and the smallest value in a dataset.
### Formula
Range=Maximum Value−Minimum Value

#### Numerical Example

Dataset:

10, 20, 30, 40, 50

Maximum = 50

Minimum = 10

Range =

50 - 10 = 40
#### Python Implementation
data = [10, 20, 30, 40, 50]

range_value = max(data) - min(data)

print("Range:", range_value)
Output
Range: 40
#### Business Example

Monthly Sales:"

₹25,000
₹40,000
₹30,000
₹55,000
₹35,000

Range tells the difference between the highest and lowest sales.

#### AI/ML Example

Student exam scores:

40, 50, 60, 90

Range helps identify how much variation exists among student scores.

Interpretation
Small Range → Data is close together.
Large Range → Data is widely spread.

## 2. Variance
### Definition

Variance measures the average squared distance of each data point from the mean.

A higher variance means the data is more spread out.



#### Numerical Example

Dataset:

2, 4, 6

Mean

= (2+4+6)/3
= 4

Squared differences

(2-4)² = 4

(4-4)² = 0

(6-4)² = 4

Variance

(4+0+4)/3

= 2.67
#### Python Implementation
import numpy as np

data = [2, 4, 6]

print("Variance:", np.var(data))
Output
Variance: 2.6666666666666665
Business Example

Variance helps companies understand fluctuations in monthly profits.

#### AI/ML Example

Variance is used in feature analysis to understand how much a feature varies.

#### Interpretation
Low Variance → Values are close to the mean.
High Variance → Values are spread out.

## 3. Standard Deviation
### Definition

Standard Deviation is the square root of variance.

It measures the average distance of data points from the mean in the original unit.


#### Numerical Example

Variance = 16

Standard Deviation

√16 = 4
#### Python Implementation
import numpy as np

data = [2,4,6]

print("Standard Deviation:", np.std(data))
Output
Standard Deviation: 1.632993161855452
#### Business Example

A company measures variation in employee salaries.

#### AI/ML Example

Standardization uses standard deviation:

Z = (X - Mean) / Standard Deviation
#### Interpretation
Low SD → Consistent data.
High SD → Data varies greatly.

## 4. Quartiles
### Definition

Quartiles divide the dataset into four equal parts.

There are three quartiles:

Q1 → 25%
Q2 → 50% (Median)
Q3 → 75%

#### Numerical Example

Dataset

10,20,30,40,50,60,70,80
Q1 = 25

Q2 = 45

Q3 = 65
#### Python Implementation
import numpy as np

data = [10,20,30,40,50,60,70,80]

print("Q1:", np.percentile(data,25))
print("Q2:", np.percentile(data,50))
print("Q3:", np.percentile(data,75))
Output
Q1: 27.5

Q2: 45.0

Q3: 62.5
#### Business Example

Quartiles divide customer spending into four groups.

AI/ML Example

Quartiles help detect outliers before model training.

#### Interpretation

Quartiles help understand how the data is distributed.

## 5. Percentiles
### Definition

Percentiles divide data into 100 equal parts.

The Pth percentile indicates that P% of the data falls below that value.

#### Numerical Example

Dataset

10,20,30,40,50

50th Percentile

= 30
#### Python Implementation
import numpy as np

data = [10,20,30,40,50]

print("90th Percentile:", np.percentile(data,90))
Output
90th Percentile: 46.0
#### Business Example

Salary reports often use the 90th percentile.

AI/ML Example

Percentiles help identify extreme values before training models.

#### Interpretation

Higher percentile means a higher relative position in the dataset.

## 6. Interquartile Range (IQR)
Definition

IQR measures the spread of the middle 50% of the data.
### Formula
IQR=Q3−Q1

#### Numerical Example
Q1 = 25

Q3 = 65

IQR = 40
####  Python Implementation
import numpy as np

data = [10,20,30,40,50,60,70,80]

q1 = np.percentile(data,25)
q3 = np.percentile(data,75)

print("IQR:", q3-q1)
Output
IQR: 35.0
#### Business Example

IQR helps detect unusually high or low sales.

####  AI/ML Example

IQR is widely used for outlier detection.

Outlier Rule

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR
####  Interpretation

Smaller IQR means the middle half of the data is closely packed.

## 7. Coefficient of Variation (CV)
### Definition

Coefficient of Variation compares the amount of variation relative to the mean.

It is useful when comparing datasets with different units or scales.


#### Numerical Example

Mean

50

Standard Deviation

10

CV

(10/50) × 100

= 20%
##### Python Implementation
import numpy as np

data = [20,30,40,50,60]

mean = np.mean(data)
std = np.std(data)

cv = (std / mean) * 100

print("Coefficient of Variation:", cv)
Output
Coefficient of Variation: 35.35533905932738
##### Business Example

Compare the consistency of returns from two investment portfolios.

##### AI/ML Example

CV helps compare the variability of different features measured on different scales.

##### Interpretation
Lower CV → More consistent data.
Higher CV → More variability.

| Measure                  | Purpose                                     | Formula           |
| ------------------------ | ------------------------------------------- | ----------------- |
| Range                    | Difference between highest and lowest value | Max − Min         |
| Variance                 | Average squared deviation from the mean     | Σ(x−μ)² / N       |
| Standard Deviation       | Average spread in original units            | √Variance         |
| Quartiles                | Divide data into four equal parts           | Q1, Q2, Q3        |
| Percentiles              | Divide data into 100 equal parts            | Pth Percentile    |
| IQR                      | Spread of the middle 50%                    | Q3 − Q1           |
| Coefficient of Variation | Relative variability                        | (SD / Mean) × 100 |
