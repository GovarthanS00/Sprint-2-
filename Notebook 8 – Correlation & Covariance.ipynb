## 1. Correlation
### Definition

Correlation measures how strongly and in what direction two variables are related.

The correlation value ranges from -1 to +1.

+1 → Perfect Positive Correlation

0 → No Correlation

-1 → Perfect Negative Correlation

```python
import pandas as pd

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [35, 45, 55, 65, 75]
}

df = pd.DataFrame(data)

print(df.corr())
```

### Business Example

More advertising budget generally leads to higher sales.

### AI/ML Example

Check whether Age and Salary are correlated with the target variable before training a model.

### Key Points
Measures relationship between two variables

Values range from -1 to +1

Does not imply causation

Used in feature analysis and EDA

## 2. Covariance
### Definition

Covariance measures whether two variables change together.

* Positive → Variables increase together
* Negative → One increases while the other decreases

Unlike correlation, covariance has no fixed range.

```python
import pandas as pd

df = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": [2, 4, 6, 8]
})

print(df.cov())
```

### Business Example

Advertising cost and sales revenue.

### AI/ML Example

Used while calculating Principal Component Analysis (PCA).

### Key Points
* Shows direction of relationship
* No fixed range
* Depends on data scale

## 3. Positive Correlation
### Definition

When one variable increases, the other also increases.

```python
import pandas as pd

df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [30, 40, 50, 60, 70]
})

print(df.corr())
```

### Business Example

Experience increases → Salary increases.

### AI/ML Example

More training data generally improves model performance.

### Key Points
* Variables move in the same direction
* Correlation value is positive

## 4. Negative Correlation
### Definition

When one variable increases, the other decreases.

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 90, 80, 70],
    "Demand": [20, 30, 40, 50]
})

print(df.corr())
```

####  Business Example

Price increases → Demand decreases.

#### AI/ML Example

Higher regularization generally reduces model complexity.

### Key Points
* Variables move in opposite directions
* Correlation value is negative

## 5. No Correlation
### Definition

There is no relationship between two variables.

```python
import numpy as np
import pandas as pd

np.random.seed(10)

df = pd.DataFrame({
    "A": np.random.randint(1, 100, 20),
    "B": np.random.randint(1, 100, 20)
})

print(df.corr())
```

### Business Example

Employee ID and salary.

### AI/ML Example

A random noise feature has no relationship with the target variable.

### Key Points
* Variables are unrelated
* Correlation is close to zero

## 6. Pearson Correlation
### Definition

Pearson Correlation measures the linear relationship between two numerical variables.

```python
import pandas as pd

df = pd.DataFrame({
    "Height": [150, 160, 170, 180],
    "Weight": [50, 60, 70, 80]
})

print(df.corr(method="pearson"))
```

### Business Example

Salary and years of experience.

### AI/ML Example

Used for feature selection before Linear Regression.

### Key Points
* Measures linear relationships
* Most commonly used correlation method
* Sensitive to outliers

## 7. Spearman Correlation
### Definition

Spearman Correlation measures the relationship between the ranks of two variables.

It works well even when the relationship is not perfectly linear.

### Business Example

Customer satisfaction rankings.

### AI/ML Example

Comparing feature rankings during feature selection.

### Key Points
* Uses ranks instead of actual values
* Works with monotonic relationships
* Less affected by outliers

## 8. Correlation Matrix
### Definition

A Correlation Matrix is a table that shows the correlation between every pair of variables in a dataset.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "Salary": [30000, 40000, 50000, 60000],
    "Experience": [1, 3, 5, 7]
})

print(df.corr())
```

### Business Example

Analyze relationships among Sales, Profit, and Advertising.

### AI/ML Example

Detect highly correlated features before model training.

### Key Points
* Shows pairwise correlations
* Useful for feature selection
* Used before building ML models

## 9. Heatmap
### Definition

A Heatmap is a color-coded visualization of a correlation matrix.

* Dark colors → Strong correlation
* Light colors → Weak correlation

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "Salary": [30000, 40000, 50000, 60000],
    "Experience": [1, 3, 5, 7]
})

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

### Business Example

Visualize relationships between Sales, Profit, Discount, and Revenue.

### AI/ML Example

Quickly identify highly correlated features during Exploratory Data Analysis (EDA).

### Key Points
* Visual representation of a correlation matrix
* Easy to identify strong and weak relationships
* Commonly used in EDA
