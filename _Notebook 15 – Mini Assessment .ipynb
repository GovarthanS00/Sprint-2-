## Topic 4: Bayes' Theorem


### Q: Explain Bayes' Theorem with an example.

```python
# Prior Probability
P_A = 0.30

# Likelihood
P_B_given_A = 0.80

# Evidence
P_B = 0.50

# Calculate Posterior Probability
P_A_given_B = (P_B_given_A * P_A) / P_B

# Print Result
print("Posterior Probability:", P_A_given_B)
```

## Topic 6: Correlation and Covariance

### Q: Explain Correlation and Covariance.

```python
import numpy as np

# Data
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 60]

# Calculate Correlation
correlation = np.corrcoef(x, y)[0, 1]

# Calculate Covariance
covariance = np.cov(x, y)[0, 1]

# Display Results
print("Correlation:", correlation)
print("Covariance:", covariance)
```

## Topic 8: Interquartile Range (IQR)

### Q: Explain IQR.

```python
import numpy as np

# Dataset
data = [5, 7, 8, 12, 15, 18, 20, 22, 25, 30]

# Calculate Q1 (25th percentile)
Q1 = np.percentile(data, 25)

# Calculate Q3 (75th percentile)
Q3 = np.percentile(data, 75)

# Calculate IQR
IQR = Q3 - Q1

# Display Results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Topic 9: p-value

### Q: What is a p-value?

```python
from scipy.stats import ttest_1samp

# Sample data
sample = [72, 75, 78, 74, 77, 79, 80]

# Claimed population mean
population_mean = 75

# Perform One-Sample T-Test
t_statistic, p_value = ttest_1samp(sample, population_mean)

# Display results
print("T-Statistic:", t_statistic)
print("P-Value:", p_value)
```

## 12. Detect Outliers using the IQR Method

```python
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 100]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_limit or x > upper_limit]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
```
