## 1. Bernoulli Distribution
### Definition

A Bernoulli distribution models an experiment that has only two possible outcomes:

* Success (1)

* Failure (0)

Formula
P(X=x)=px(1−p)1−x

Where:
x∈{0,1}

p = probability of success

Used When
* Single trial
* Yes/No problems
* Pass/Fail
*Fraud/Not Fraud

### Real-world Example

Did a customer purchase a product?

Yes = 1
No = 0
### AI/ML Example

### Email Classification

* Spam = 1
* Not Spam = 0
* Python Visualization

```python
##Python Visualization

##Generate many Bernoulli samples and visualize using a bar chart.

##Libraries:

numpy.random.binomial(n=1, p=0.7)
```

## 2. Binomial Distribution
### Definition

A Binomial distribution models the number of successes in a fixed number of independent Bernoulli trials.



Where

* n = number of trials
* k = successes
* p = probability of success

### Real-world Example

10 customers visit a shop.
How many buy a product?

### AI/ML Example

Out of 100 predictions,
how many are correct?

### Python Visualization
numpy.random.binomial()

Plot histogram of simulated successes.

## 3. Poisson Distribution
### Definition

Poisson distribution models the number of events occurring in a fixed interval when events occur independently at a constant average rate.



## Used When

* Count data

* Rare events

* Events per unit time

## Real-world Example

Number of customers entering a bank per minute.

## AI/ML Example

Number of server failures per day.

Python Visualization
* numpy.random.poisson()

Histogram.

### Used When

* Random number generation

* Simulation

* Monte Carlo methods

### Real-world Example

Rolling a fair dice.

### AI/ML Example

Random initialization.

Random sampling.

### Python Visualization
numpy.random.uniform()

## 5. Normal Distribution (Gaussian Distribution)
### Definition

A continuous probability distribution where most values cluster around the mean and fewer values occur farther away, forming a bell-shaped curve.

### Used When

* Natural phenomena

* Statistics

*  Machine Learning

* Data Science

### Real-world Example

* Human heights

* Exam marks

* IQ scores

### AI/ML Example

Many ML algorithms assume normally distributed features.

### Examples:

* Linear Regression
* PCA
* Gaussian Naive Bayes
* Python Visualization
numpy.random.normal()

Histogram with density curve.
