## Statistics

 Statistics is the science of collecting, organizing, analyzing, and interpreting data to understand information and make better decisions.

## Example

A teacher has the marks of 5 students:

70, 75, 80, 85, 90

Instead of looking at each mark separately, the teacher calculates the average mark (80) to understand the overall class performance.

This is Statistics because it summarizes the data into useful information.

```python
import numpy as np

marks = [70, 75, 80, 85, 90]

average = np.mean(marks)

print("Student Marks:", marks)
print("Average Marks:", average)
```

## Why is Statistics important in AI/ML?

Statistics helps us understand data before building an AI/ML model. It helps identify patterns, summarize data, and make better predictions.
## Example

A bank wants to predict whether a customer will repay a loan.

Before building the ML model, it analyzes customer data such as Age, Income, and Loan Amount using statistics.

Statistics helps understand the data before making predictions.

## 3. Statistics vs Machine Learning
Statistics is used to understand and analyze existing data.
Machine Learning is used to learn patterns from data and make predictions on new data.

## Example
A company has employee salary data.

Statistics → Calculate the average salary.
Machine Learning → Predict the salary of a new employee based on experience.

This shows that Statistics analyzes data, while Machine Learning predicts future outcomes.

```python
import numpy as np

salary = [30000, 35000, 40000, 45000, 50000]

average_salary = np.mean(salary)

print("Employee Salary:", salary)
print("Average Salary:", average_salary)

print("\nMachine Learning Example:")
print("Predict the salary of a new employee based on previous salary data.")
```

## 4.Descriptive Statistics

Descriptive Statistics is used to summarize and describe data using measures such as Mean, Median, and Mode

## Example

A teacher has the marks of 5 students:

70, 75, 80, 85, 90

Instead of checking every mark individually, the teacher calculates the Mean, Median, and Mode to summarize the class performance.

This is Descriptive Statistics because it describes the existing data

✅ Interview Revision

Mean → Average value.

Median → Middle value after sorting.

Mode → Most frequently occurring value.

```python
import numpy as np
from statistics import mode

marks = [70, 75, 80, 85, 90]

print("Student Marks:", marks)

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Mode:", mode(marks))
```

## Inferential Statistics

Inferential Statistics is used to make conclusions or predictions about a population using a sample.
## Example
A company has 5,000 customers.

Instead of asking all 5,000 customers for feedback, it surveys 200 customers and predicts the opinion of all customers.

This is Inferential Statistics because a sample is used to make conclusions about the population.

```python
population = 5000
sample = 200

print("Population:", population)
print("Sample:", sample)

print("Conclusion: The sample is used to represent the population.")
```

## 6 – Population
Population is the entire group of people, objects, or data that we want to study.

## Example

A company has 5 employees.

If we consider all 5 employees for analysis, then all 5 employees represent the Population.

```python
import pandas as pd

employee = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["John", "Alice", "Bob", "Emma", "David"],
    "Age": [25, 30, 28, 35, 27],
    "Department": ["IT", "HR", "IT", "Finance", "Sales"],
    "Salary": [45000, 50000, 48000, 60000, 47000]
}

df = pd.DataFrame(employee)

print(df)
print("\nPopulation Size:", len(df))
```

## 9. Sampling

Sampling is the process of selecting a small group (sample) from the population for analysis.
## Example
A company has 5 employees.

Instead of collecting data from all 5 employees, it selects John and Bob.

This process of selecting only a few employees is called Sampling.

```python
sample = df.iloc[[0, 2]]

print(sample)
```

## ⭐ Easy way to remember
Population → Entire group.

Sample → Small group selected.

Census → Collect data from everyone.

Sampling → Process of selecting the small group.

These four concepts are connected and are often asked together in interviews.

## 16. Continuous Data
Continuous Data is numerical data that can take any value within a range, including decimal values.
## Example

Employee heights:

165.5 cm
170.2 cm
175.8 cm

Height is Continuous Data because it can have decimal values.

Note: Our employee dataset doesn't have a Height column, so we use a separate example.

```python
import pandas as pd

height = [165.5, 170.2, 175.8, 168.4, 172.6]

height_df = pd.DataFrame({"Height (cm)": height})

print(height_df)
```

## 17.Discrete Data

Discrete Data is numerical data that can take only whole numbers. It is obtained by counting.
## Example

Number of employees in each department:

IT → 2
HR → 1
Finance → 1
Sales → 1

These are Discrete Data because they are whole numbers.

```python
department_count = df["Department"].value_counts()

print(department_count)
```
