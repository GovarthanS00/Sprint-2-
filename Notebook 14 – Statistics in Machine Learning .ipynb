## 1. Data Cleaning
### Definition

Statistics helps identify incorrect, inconsistent, duplicate, or invalid data before training a machine learning model.

### Why Statistics is Used
* Detect incorrect values
* Find duplicate records
* Identify inconsistent data
* Improve data quality
### Business Example

A customer database contains duplicate customer IDs. Statistics helps identify and remove duplicates.

### AI/ML Example

Cleaning incorrect values improves model accuracy by providing high-quality training data.

### Key Points

* First step before model training
* Improves dataset quality
* Reduces noise

## 2. Missing Value Handling
### Definition

Statistics helps replace or remove missing values using statistical measures.

### Why Statistics is Used
* Mean
* Median
* Mode
* Forward/Backward Fill (for time series)
### Business Example

Some employees did not enter their salary. Replace missing salaries with the median salary.

### AI/ML Example

Most ML algorithms cannot handle missing values directly.

### Key Points
Prevents data loss
Makes dataset complete
Choice depends on data distribution

### 3. Outlier Detection
#### Definition

Statistics identifies unusual observations that are significantly different from other data.

#### Why Statistics is Used
* IQR Method
* Z-Score
* Box Plot
### Business Example

A customer purchased products worth ₹10 instead of the usual ₹10,000 due to an entry mistake.

### AI/ML Example

Removing extreme values improves model performance.

### Key Points
* Detects abnormal values
* Improves model accuracy
* Prevents biased predictions

### 4. Feature Engineering
#### Definition

Statistics helps create meaningful features from existing data.

#### Why Statistics is Used
* Correlation analysis
* Aggregation
* Statistical summaries
* Encoding categorical data
#### Business Example

Calculate customer's average monthly spending.

### AI/ML Example

Create "Age Group" from Age.

### Key Points
* Improves feature quality
* Helps models learn better
* Increases predictive performance

### 5. Feature Scaling
#### Definition

Statistics scales numerical features so they have comparable ranges.

#### Why Statistics is Used
* Mean
* Standard Deviation
* Min
* Max
### Business Example

Salary ranges from 20,000–200,000 while age ranges from 20–60.

### AI/ML Example

Required for KNN, SVM, Neural Networks, PCA.

### Key Points
* Prevents dominance of large values
* Faster convergence
* Better model performance

### 6. Data Normalization
#### Definition

Statistics transforms data into a common scale.

### Why Statistics is Used
* Min-Max Scaling
* Z-Score Standardization
### Business Example

Convert different product prices into the same scale.

### AI/ML Example

Improves distance-based algorithms like KNN.

### Key Points
* Common data scale
* Better optimization
* Useful for many ML algorithms

### 7. Model Evaluation
### Definition

Statistics measures how well a machine learning model performs.

### Why Statistics is Used
* Accuracy
* Precision
* Recall
* F1 Score
* RMSE
* MAE
### Business Example

Measure fraud detection accuracy.

### AI/ML Example

Compare multiple models and select the best one.

### Key Points
* Measures model performance
* Helps compare models
* Prevents poor model selection

## 8. Linear Regression
### Definition

Statistics finds the relationship between independent and dependent variables.

### Why Statistics is Used
* Correlation
* Variance
* Residual Analysis
### Business Example

Predict house prices based on area.

### AI/ML Example

Predict employee salary from years of experience.

### Key Points
* Predicts continuous values
* Assumes linear relationship
* Uses statistical estimation

### 9. Logistic Regression
#### Definition

Statistics estimates the probability of a categorical outcome.

### Why Statistics is Used
* Probability
* Maximum Likelihood Estimation
### Business Example

Predict whether a customer will buy a product.

### AI/ML Example

Spam email detection.

### Key Points
* Binary classification
* Outputs probabilities
* Widely used baseline classifier

### 10. Naive Bayes
#### Definition

A probabilistic algorithm based on Bayes' Theorem.

### Why Statistics is Used
* Conditional Probability
* Probability Distribution
### Business Example

Classify customer reviews as positive or negative.

### AI/ML Example

Email spam filtering.

### Key Points
* Fast algorithm
* Works well with text data
* Assumes feature independence

### 11. Clustering
#### Definition

Statistics groups similar data points without labels.

### Why Statistics is Used
* Distance Measures
* Mean (Centroid)
* Variance
### Business Example

### AI/ML Example

Customer segmentation using K-Means.

### Key Points
* Unsupervised learning
* Finds hidden patterns
* Used for segmentation

### 12. Principal Component Analysis (PCA)
#### Definition

Statistics reduces the number of features while preserving most of the information.

### Why Statistics is Used
* Variance
* Covariance
* Eigenvalues
### Business Example

Reduce hundreds of customer attributes into a few important features.

### AI/ML Example

Reduce dimensionality before model training.

### Key Points
* Feature reduction
* Removes redundancy
* Faster model training

### 13. Recommendation Systems
#### Definition

Statistics identifies user preferences to recommend relevant items.

### Why Statistics is Used
* Similarity Measures
* Correlation
* Probability
* Matrix Factorization
### Business Example

### AI/ML Example

### Key Points

* Improves user experience
