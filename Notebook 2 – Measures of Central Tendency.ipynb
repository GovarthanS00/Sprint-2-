## 1. Mean (Arithmetic Mean)
## Definition

Mean is the average of all values in a dataset.

Formula
Mean=
N
∑X
	​


Where

ΣX = Sum of all values
N = Number of value

#### Numerical Example

Dataset

10, 20, 30, 40, 50

Calculation

Mean = (10 + 20 + 30 + 40 + 50) / 5

#### Python Implementation
import numpy as np

## 4. Weighted Mean
### Definition

Weighted Mean gives different importance (weights) to values.


#### Numerical Example
| Subject | Marks | Credits |
| ------- | ----- | ------- |
| Math    | 80    | 4       |
| Science | 90    | 3       |
| English | 70    | 2       |

(80×4 + 90×3 + 70×2)

---------------------

(4+3+2)

=730/9

=81.11



```python
#### Python Implementation
import numpy as np

marks = [80,90,70]
weights = [4,3,2]

weighted_mean = np.average(marks, weights=weights)

print(weighted_mean)
```

#### Business Example

Customer Satisfaction

Premium customers have higher weights than regular customers.

#### AI/ML Example

Weighted Loss Functions

More important classes receive higher weights during training.

#### Interpretation
Useful when observations have different importance.
More realistic than simple average in many cases.

## 5. Geometric Mean
### Definition

Geometric Mean is the nth root of the product of all values.

Useful for growth rates.



#### Numerical Example
2 8

Calculation

√16

=4
#### Python Implementation
from scipy.stats import gmean

data = [2,8]
print(gmean(data))
Output
4.0
#### Business Example

Investment returns

10%

20%

15%

Average growth is better measured using Geometric Mean.

#### AI/ML Example

Useful when working with

Growth ratios
Probability values
Log-transformed data
Interpretation
Best for percentages.
Best for compound growth.
Cannot contain zero or negative values.

## 6. Harmonic Mean
### Definition

Harmonic Mean is the reciprocal of the average of reciprocals.

Best for rates.



#### Numerical Example

Speeds

60 km/hr

40 km/hr

Calculation

HM

=2

/(1/60 + 1/40)

=48
#### Python Implementation
from scipy.stats import hmean

data = [60,40]

print(hmean(data))
Output
48.0
####  Business Example

Average production rate

Workers produce items at different speeds.

#### AI/ML Example

Used in

F1 Score
F1=2×
Precision+Recall
Precision×Recall
F1 Score is the harmonic mean of Precision and Recall.

#### Interpretation
Best for averages of rates.
Gives more importance to smaller values.
Used in ML evaluation metrics.
