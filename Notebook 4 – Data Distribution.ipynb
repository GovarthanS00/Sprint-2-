## Notebook 4 – Data Distribution
### 1. Normal Distribution
#### Definition

A Normal Distribution is a continuous probability distribution where data is symmetrically distributed around the mean.

Characteristics

Bell-shaped curve

Mean = Median = Mode

Symmetric

Total Area = 1 (100%)



```python
## Python Implementation
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 5, 1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.show()
```

Student Marks

48, 50, 52, 49, 51, 50, 53

Most values are close to 50.

### Business Example

Employee salaries in a large company often follow a normal distribution.

### AI/ML Example

Many ML algorithms assume normally distributed features.

### Interpretation
Data is balanced.
Extreme values are rare.
Most observations lie near the mean.

## 2. Uniform Distribution
### 1. Definition

A Uniform Distribution is a distribution where every value has an equal chance of occurring.

The graph looks flat, as no value is more likely than another.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 values between 0 and 10
data = np.random.uniform(low=0, high=10, size=1000)

# Plot histogram
plt.hist(data, bins=20)

# Add title and labels
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

## 3. Business Example 🏢

Online Shopping Coupon

An e-commerce company generates coupon numbers.

Every customer should have an equal chance of receiving any coupon number.

No customer should get priority over another.

This follows a Uniform Distribution.

## 4. AI/ML Example 🤖

Customer Churn Prediction

Before training the model, the dataset is randomly split into:

80% Training Data
20% Testing Data

Every customer record should have an equal chance of being selected.

This prevents bias and helps the model learn fairly.

## 5. Key Points
Every value has an equal probability.
Graph looks flat.
Business: Random coupon generation.
AI/ML: Random sampling, data shuffling, train-test split.

## 3. Gaussian Distribution
### 1. Definition

A Gaussian Distribution is another name for the Normal Distribution.

There is no difference between them. The name "Gaussian" comes from the mathematician Carl Friedrich Gauss.

The graph is bell-shaped, where most values are near the average.

### Interview Tip:
Normal Distribution = Gaussian Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 values using Gaussian (Normal) Distribution
data = np.random.normal(loc=50, scale=5, size=1000)

# Plot histogram
plt.hist(data, bins=30)

# Add title and labels
plt.title("Gaussian Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

### Note:
We use np.random.normal() because NumPy uses the term "normal", even though it is the same as a Gaussian Distribution.

#### 3. Business Example 🏢
Manufacturing Quality Check

A company manufactures 1-liter water bottles.

Most bottles contain 1 liter.
A few contain 0.98 liters.
A few contain 1.02 liters.

The measurements form a Gaussian (Normal) Distribution.

This helps the company monitor product quality.

####  4. AI/ML Example 🤖
Image Classification

You're building an AI model to classify cats and dogs.

Before training, you normalize the image pixel values.

Many AI/ML techniques assume the data is approximately Gaussian (normally distributed), which can help models train more effectively.

You'll frequently hear terms like:

Gaussian Noise
Gaussian Naive Bayes
Gaussian Blur

All of these are based on the Gaussian Distribution.

####  5. Key Points
Gaussian Distribution = Normal Distribution.
Bell-shaped graph.
Most values are near the average.
Business: Product quality measurement.
AI/ML: Used in Gaussian Naive Bayes, Gaussian Noise, image processing, and data preprocessing.

## 4. Left Skew (Negative Skew)
### 1. Definition

A Left Skew Distribution is a distribution where most values are on the right side, and a few very small values create a long tail on the left side.

### Simply remember:

Tail points to the Left = Left Skew

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 values with a left-skewed distribution
data = np.random.beta(a=5, b=2, size=1000)

# Plot histogram
plt.hist(data, bins=20)

# Add title and labels
plt.title("Left Skew Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

## 5. Right Skew (Positive Skew)
### 1. Definition

A Right Skew Distribution is a distribution where most values are on the left side, and a few very large values create a long tail on the right side.

Easy Rule:

Tail points to the Right = Right Skew

```python
import matplotlib.pyplot as plt

# Customer purchase amounts (₹)
purchase = [200, 250, 300, 350, 400, 450, 500, 550, 3000]

# Plot histogram
plt.hist(purchase, bins=5)

# Add title and labels
plt.title("Right Skew Distribution")
plt.xlabel("Purchase Amount (₹)")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

### What this code does
Creates a list of customer purchase amounts.

Most customers spend between ₹200–₹550.

One customer spends ₹3000.

The graph will show a tail on the right side.

### 6. Symmetric Distribution
## 1. Definition

A Symmetric Distribution is a distribution where both the left and right sides are almost mirror images of each other.

There is no left skew and no right skew.

Think of folding the graph in the middle—both sides look almost the same.

```python
import matplotlib.pyplot as plt

# Student marks
marks = [40, 45, 48, 50, 52, 55, 60]

# Plot histogram
plt.hist(marks, bins=5)

# Add title and labels
plt.title("Symmetric Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

### What this code does
Creates a list of student marks.

The marks are balanced around 50.

The histogram will look nearly the same on both sid

## 7. Kurtosis
### 1. Definition

Kurtosis tells us whether a dataset has more or fewer extreme values (outliers) compared to a normal distribution.

In simple words:

Kurtosis helps us understand how many unusually high or low values are present in the data.

Don't think of it as "peak height." For AI/ML, it's more useful to think about outliers

```python
import pandas as pd

# Employee salaries (₹ in Lakhs)
salary = [4, 5, 5, 6, 6, 6, 7, 7, 8, 30]

# Create DataFrame
df = pd.DataFrame({"Salary": salary})

# Calculate kurtosis
print(df["Salary"].kurt())
```

### 3. Business Example 🏢
Company Salary Analysis

A company has:

Most employees earning ₹5–8 Lakhs
One CEO earning ₹2 Crores

That CEO's salary is an outlier.

Kurtosis helps identify that the salary distribution contains extreme values.

    High kurtosis → More extreme values.

        Low kurtosis → Fewer extreme values.

## 8. Empirical Rule (68–95–99.7 Rule)
### 1. Definition

The Empirical Rule applies to Normal Distribution.

It tells us where most of the data lies.

Simply remember:

68% of data is close to the average.
95% of data is a little farther from the average.
99.7% of data is almost the entire dataset.

This rule works only for data that follows a Normal Distribution.

```python
import numpy as np

# Student marks
marks = [45, 48, 50, 52, 49, 51, 53, 47, 50, 49]

# Calculate average and standard deviation
mean = np.mean(marks)
std = np.std(marks)

print("Mean :", mean)
print("Standard Deviation :", std)

print("68% Range :", mean - std, "to", mean + std)
print("95% Range :", mean - (2 * std), "to", mean + (2 * std))
print("99.7% Range :", mean - (3 * std), "to", mean + (3 * std))
```

### What this code does
Calculates the average (mean).
Calculates the standard deviation (how spread out the data is).
Shows the ranges where approximately 68%, 95%, and 99.7% of the data are expected to fall

### 3. Business Example 🏢
Manufacturing Company

A factory produces 1-liter milk packets.

Most packets contain about 1 liter.

If one packet contains only 0.5 liters, it's unusual.

Using the Empirical Rule, the company can quickly identify products that are far from the normal range and inspect them.

### 4. AI/ML Example 🤖
Credit Card Fraud Detection

You're building a fraud detection model.

Most customer transactions are between:

₹500 – ₹2,000

Suddenly, one transaction is:

₹8,00,000

This transaction is far outside the normal range.

Using the Empirical Rule, the AI system can flag it as an anomaly for further investigation.

This doesn't automatically mean it's fraud—it means it's unusual and deserves attention.

### 5. Key Points
Works only for Normal Distribution.
68% of data is close to the average.
95% includes almost all common values.
99.7% includes nearly the entire dataset.
Business: Quality control.
AI/ML: Anomaly detection and fraud detection.
