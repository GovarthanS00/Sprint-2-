## 1. Histogram
### Purpose

A histogram shows the distribution of numerical data by grouping values into continuous intervals (bins).

### When to Use
* To understand data distribution
* To detect skewness
* To identify outliers
* To check whether data follows a normal distribution
### Advantages
* Easy to understand data distribution
* Shows frequency of values
* Helps identify patterns and outliers
### Limitations
* Choice of bin size affects the visualization
* Not suitable for categorical data
  ### Interpretation
* Taller bars indicate more observations in that range.
* A symmetric shape suggests a normal distribution.
* Long tails indicate skewness.

```python
import matplotlib.pyplot as plt
import numpy as np

salary = np.random.normal(50000, 10000, 500)

plt.hist(salary, bins=15)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

## 2. Bar Chart
### Purpose

A bar chart compares values across different categories.

### When to Use
* Comparing categories
* Showing counts or totals
* Displaying survey results
### Advantages
* Very easy to read
* Good for categorical data
* Highlights differences clearly
### Limitations
* Not suitable for continuous numerical distributions
* Too many categories make it cluttered

### Interpretation
Taller bars represent higher values.
Easily compare different categories.

```python
import matplotlib.pyplot as plt

departments = ["HR", "IT", "Sales", "Finance"]
employees = [15, 40, 28, 18]

plt.bar(departments, employees)
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Employees")
plt.show()
```

## 3. Line Chart
### Purpose

Shows how values change over time or in a sequence.

### When to Use
* Time-series data
* Trend analysis
* Sales growth
* Stock prices
### Advantages
* Clearly shows trends
* Easy to identify increases and decreases
### Limitations
* Not suitable for unrelated categories
* Multiple lines may become confusing
### Interpretation
Upward trend indicates growth.
Downward trend indicates decline.
Flat line indicates stability.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 180, 170, 210]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

### 4. Scatter Plot
#### Purpose

Shows the relationship between two numerical variables.

### When to Use
* Correlation analysis
* Detecting outliers
* Feature relationship analysis
### Advantages
* Shows patterns clearly
* Helps identify clusters
* Detects outliers
### Limitations
* Not useful for categorical variables
* Large datasets may overlap
### Interpretation
* Upward pattern indicates positive correlation.
* Downward pattern indicates negative correlation.
* Random spread indicates little or no correlation.

```python
import matplotlib.pyplot as plt
import numpy as np

experience = np.random.randint(1, 15, 100)
salary = experience * 5000 + np.random.randint(-8000, 8000, 100)

plt.scatter(experience, salary)
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

## 5. Pie Chart
### Purpose

Shows the proportion of each category in a whole.

### When to Use
* Percentage distribution
* Market share
* Budget allocation
### Advantages
* Easy to understand percentages
* Good for few categories
### Limitations
* Difficult to compare many categories
* Not suitable for large datasets
### Interpretation
* Larger slices represent larger proportions.
* Total always equals 100%.

```python
import matplotlib.pyplot as plt

labels = ["HR", "IT", "Sales", "Finance"]
sizes = [15, 40, 30, 15]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()
```

## 6. Density Plot (KDE Plot)
### Purpose

Shows the probability density of continuous data using a smooth curve.

### When to Use
* Understanding data distribution
* Comparing multiple distributions
* Checking normality
### Advantages
* Smooth representation
* Easier than histogram for comparing distributions
### Limitations
* Depends on bandwidth selection
* May hide small details
### Interpretation
* Higher peaks indicate more common values.
* Wider curves indicate greater variability.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

salary = np.random.normal(50000, 10000, 500)

sns.kdeplot(salary)
plt.title("Density Plot")
plt.show()
```

## 7. Box Plot
### Purpose

Shows data distribution using quartiles and identifies outliers.

### When to Use
* Detecting outliers
* Comparing multiple groups
* Understanding spread
### Advantages
* Quickly identifies outliers
* Shows median and spread
* Easy group comparison
### Limitations
* Does not show the exact distribution shape
* Less intuitive for beginners
### Interpretation
* Middle line = Median
* Box = Middle 50% of data (IQR)
* Points outside whiskers = Outliers

```python
import matplotlib.pyplot as plt
import numpy as np

salary = np.random.normal(50000, 10000, 200)

plt.boxplot(salary)
plt.title("Salary Box Plot")
plt.show()
```

## 8. Violin Plot
### Purpose

Combines a box plot with a density plot to show both summary statistics and data distribution.

### When to Use
* Comparing distributions across categories
* Understanding density and spread
### Advantages
* Shows full distribution
* Displays density and median together
### Limitations
* Slightly harder to interpret
* May be unnecessary for small datasets
### Interpretation
* Wider sections indicate more observations.
* Narrow sections indicate fewer observations.
* Inner box shows the median and quartiles.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.violinplot(x="day", y="total_bill", data=tips)
plt.title("Violin Plot")
plt.show()
```

## 9. Pair Plot
### Purpose

Displays pairwise relationships between numerical features.

### When to Use
* Exploratory Data Analysis (EDA)
* Feature selection
* Detecting correlations
### Advantages
* Visualizes multiple relationships at once
* Helps detect patterns and clusters
### Limitations
* Slow for large datasets
* Difficult with many features
### Interpretation
* Diagonal plots show each feature's distribution.
* Off-diagonal plots show relationships between feature pairs.
* Clusters may indicate different classes.

```python
import seaborn as sns

iris = sns.load_dataset("iris")
sns.pairplot(iris)
```

## 10. Heatmap
### Purpose

Displays values as colors, commonly used to visualize correlation matrices.

### When to Use
* Correlation analysis
* Missing value visualization
* Comparing multiple variables
### Advantages
* Easy to identify strong relationships
* Useful for feature selection
* Color-coded for quick interpretation
### Limitations
* Too many variables reduce readability
* Exact values may require annotations
### Interpretation
* Values close to 1 indicate a strong positive correlation.
* Values close to -1 indicate a strong negative correlation.
( Values close to 0 indicate little or no correlation.

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

corr = iris.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
