## 1. What are Outliers?
### Definition

An outlier is a data point that is significantly different from the rest of the observations in a dataset. It can be much larger or much smaller than most values.

### Outliers may occur due to:

* Data entry errors
* Measurement errors
* Natural variation
* Rare events

### Why are Outliers Important?
1. Can affect the mean and standard deviation.
2. Can reduce machine learning model accuracy.
3. May indicate fraud or unusual behavior.
4. Need to be detected before training models.

```python
import numpy as np

data = np.array([10, 12, 13, 15, 16, 18, 20, 120])

print(data)
```

### Business Example

A customer's monthly purchase amount is usually between $100 and $500, but one customer spends $15,000.

This purchase is an outlier.

### AI/ML Example

In a house price dataset, most homes cost between $200,000 and $600,000, but one mansion costs $8,000,000.

This extremely high price is an outlier and can influence regression models.

### Key Points
1. Much higher or lower than other observations.
2. Can distort statistical analysis.
3. Should be investigated before removing.

### 2. Types of Outliers

There are three common types of outliers.

A. Global Outlier
B. Contextual Outlier
C. Collective 

### A. Global Outlier
### Definition

A value that is far away from all other data points.

### Example
10 11 12 13 14 15 120

Here, 120 is a global outlier.

### B. Contextual Outlier
### Definition

A value that is unusual only in a specific context.

### Example

Temperature of 20°C

Normal in winter
Outlier in summer

### C. Collective Outlier
### Definition

A group of observations that together behave abnormally.

### Example

A server suddenly shows high CPU usage continuously for one hour.

Each reading may look normal individually, but together they form an abnormal pattern.

```python
import numpy as np

data = np.array([10, 12, 13, 15, 18, 20, 150])

print(data)
```

150 is a global outlier.

### Business Example

Credit card fraud detection.

One transaction of $20,000 while the customer usually spends $50.

### AI/ML Example

Fraud detection systems classify abnormal transactions as outliers.

### 3. Detecting Outliers

Several techniques are used to detect outliers

| Method       | Best For                  |
| ------------ | ------------------------- |
| IQR Method   | Skewed data               |
| Z-Score      | Normally distributed data |
| Box Plot     | Visualization             |
| Scatter Plot | Visual detection          |


### Business Example

Banks detect unusual transactions before approving payments.

### AI/ML Example

Outlier detection is often performed during data preprocessing before model training.

### Key Points
Detect before training ML models.
Different methods suit different data distributions.

### 4. IQR Method
### Definition

The Interquartile Range (IQR) measures the spread of the middle 50% of the data. Values outside the lower and upper limits are considered outliers.

### Formula
* IQR = Q3 − Q1
* Lower Bound = Q1 − 1.5 × IQR
* Upper Bound = Q3 + 1.5 × IQR

```python
import pandas as pd

data = pd.Series([10,12,13,15,16,18,20,120])

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print(outliers)
```

### Business Example

Detect customers with unusually high spending.

### AI/ML Example

Remove extreme values before training Linear Regression.

### Key Points
* Works well for skewed data.
* Does not assume normal distribution.
* Widely used in preprocessing.

## 6. Box Plot
### Definition

A Box Plot is a graphical representation of data that shows:

* Minimum
* Q1
* Median
* Q3
* Maximum
* Outliers

Outliers appear as individual points outside the whiskers.

```python
import matplotlib.pyplot as plt

data = [10,12,13,15,16,18,20,120]

plt.boxplot(data)

plt.title("Box Plot")

plt.show()
```

###  Business Example

Visualize employee salary distribution.

### AI/ML Example

Quickly identify extreme feature values before preprocessing.

### Key Points
* Easy to identify outliers.
* Based on the IQR method.
* Commonly used in Exploratory Data Analysis (EDA).

## 7. Scatter Plot
### Definition

A Scatter Plot displays data points on two axes. Outliers appear as points far away from the main cluster.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6,7,8]
y = [10,12,13,15,16,18,20,120]

plt.scatter(x, y)

plt.title("Scatter Plot")

plt.xlabel("X")

plt.ylabel("Y")

plt.show()
```

### Business Example

Identify customers whose spending is unusually high compared to others.

### AI/ML Example

Detect abnormal sensor readings before training predictive models.

### Key Points
* Useful for visual outlier detection.
* Shows relationships between variables.
* Helps identify unusual observations quickly.
