## Statistical Tests

Statistical tests help determine whether the observed results in data are due to chance or represent a real difference or relationship.

They are widely used in:

* Data Analysis
* Business Decision Making
* A/B Testing
* Machine Learning
* Scientific Research

## 1. T-Test
### Definition

A T-Test is used to determine whether the means of two groups are significantly different when the sample size is small or the population standard deviation is unknown.

### Assumptions
* Data is continuous.
* Data is approximately normally distributed.
* Samples are randomly selected.
* Independent samples (for Independent T-Test).
* Equal variance (for Student's T-Test).
### When to Use

Use a T-Test when:

* Comparing two groups.
* Sample size is small (typically <30).
* Population standard deviation is unknown.

### Examples:

* Compare salaries of two departments.
* Compare model accuracy before and after tuning.
* When Not to Use

### Do not use when:

* Comparing more than two groups (use ANOVA).
* Data is categorical.
* Data is highly non-normal (use Mann-Whitney U Test).
### Business Example

A company wants to know whether the average monthly sales of Store A and Store B are different.

```python
from scipy.stats import ttest_ind

group1 = [50,55,60,62,58]
group2 = [45,48,50,52,47]

t_stat, p_value = ttest_ind(group1, group2)

print("T Statistic:", t_stat)
print("P Value:", p_value)
```

## 2. Z-Test
### Definition

A Z-Test compares sample means when the sample size is large or the population standard deviation is known.

### Assumptions
* Large sample size (usually ≥30)
* Normal distribution
* Population standard deviation is known
* Random sampling
### When to Use

Use a Z-Test when:

* Large datasets
* Known population standard deviation
* Comparing means

### When Not to Use

Do not use when:

* Small sample size
* Population standard deviation is unknown

### Business Example

A factory checks whether the average weight of packaged products matches the expected weight.

```python
from statsmodels.stats.weightstats import ztest

data = [52,55,51,53,54,56,52,51,55,54]

z_stat, p_value = ztest(data, value=50)

print("Z Statistic:", z_stat)
print("P Value:", p_value)
```

### Interpretation
p-value < 0.05 → Reject the null hypothesis.
p-value ≥ 0.05 → Fail to reject the null hypothesis.
### AI/ML Example

Compare prediction errors of a large number of model predictions against a benchmark.

## 3. Chi-Square Test
### Definition

A Chi-Square Test determines whether there is a significant relationship between categorical variables.

### Assumptions
* Categorical data
* Independent observations
* Expected frequency in each cell is usually at least 5
### When to Use

Use when checking association between categories.

### Examples:

* Gender vs Product Preference
* Education vs Job Selection

### When Not to Use

Do not use when:

* Data is numerical.
* Expected frequencies are too small.
  
### Business Example

Determine whether customer gender influences product purchases.

```python
from scipy.stats import chi2_contingency

table = [
    [40,10],
    [30,20]
]

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square:", chi2)
print("P Value:", p)
```

### Interpretation
p-value < 0.05 → Variables are associated.
p-value ≥ 0.05 → Variables are independent.
### AI/ML Example

Check whether class labels depend on a categorical feature before feature selection.

### 4. ANOVA (Analysis of Variance)
##### Definition

ANOVA compares the means of three or more groups to determine whether at least one group mean differs significantly.

### Assumptions
* Normal distribution
* Independent observations
* Equal variances
* Continuous data
* When to Use

Use when comparing three or more groups.

### Examples:

* Sales in three regions
* Scores from three teaching methods

### When Not to Use

Do not use when:

* Comparing only two groups (use T-Test)
* Data is categorical
* Assumptions are violated
* Business Example

Compare average monthly revenue across three company branches.

```python
from scipy.stats import f_oneway

group1 = [10,12,11,13]
group2 = [18,20,19,17]
group3 = [25,27,26,24]

f_stat, p_value = f_oneway(group1, group2, group3)

print("F Statistic:", f_stat)
print("P Value:", p_value)
```

### Interpretation
p-value < 0.05 → At least one group is significantly different.
p-value ≥ 0.05 → No significant difference.
### AI/ML Example

Compare model accuracy across three different algorithms.

### 5. Mann-Whitney U Test
### Definition

The Mann-Whitney U Test compares two independent groups when the data is not normally distributed.

### Assumptions
* Independent samples
* Ordinal or continuous data
* Non-normal distribution
### When to Use

Use when:

* Comparing two groups
* Data is skewed
* Normality assumption is violated
When Not to Use

Do not use when:

* Data is normally distributed (use T-Test)
* Samples are paired
### Business Example

Compare customer satisfaction ratings from two stores.

```python
from scipy.stats import mannwhitneyu

group1 = [5,7,8,6,9]
group2 = [2,3,4,5,3]

u_stat, p_value = mannwhitneyu(group1, group2)

print("U Statistic:", u_stat)
print("P Value:", p_value) 
```

### Interpretation
p-value < 0.05 → Groups differ significantly.
p-value ≥ 0.05 → No significant difference.
### AI/ML Example

Compare model performance scores when evaluation metrics are not normally distributed.

## 6. Shapiro-Wilk Test
### Definition

The Shapiro-Wilk Test checks whether a dataset follows a normal distribution.

### Assumptions
* Continuous data
* Random sample
### When to Use

Use before selecting a statistical test to verify the normality assumption.

### When Not to Use

Do not use when:

* Data is categorical.
* Sample size is extremely large (other normality tests or plots may be more appropriate).
### Business Example

Check whether employee salaries are normally distributed before performing a T-Test.

```python
from scipy.stats import shapiro

data = [12,15,14,13,16,18,17]

stat, p_value = shapiro(data)

print("Statistic:", stat)
print("P Value:", p_value)
```

### Interpretation
* p-value > 0.05 → Data is normally distributed.
* p-value ≤ 0.05 → Data is not normally distributed.
### AI/ML Example

Verify feature normality before applying algorithms or statistical methods that assume normally distributed data.

| Test           | Data Type         | Purpose            | Groups     | Parametric |
| -------------- | ----------------- | ------------------ | ---------- | ---------- |
| T-Test         | Numerical         | Compare means      | 2          | ✅ Yes      |
| Z-Test         | Numerical         | Compare means      | 2          | ✅ Yes      |
| Chi-Square     | Categorical       | Association        | Categories | ❌ No       |
| ANOVA          | Numerical         | Compare means      | 3+         | ✅ Yes      |
| Mann-Whitney U | Numerical/Ordinal | Compare two groups | 2          | ❌ No       |
| Shapiro-Wilk   | Numerical         | Test normality     | 1 Dataset  | N/A        |


1. T-Test → Compare 2 means (small sample or unknown population standard deviation).
2. Z-Test → Compare 2 means (large sample or known population standard deviation).
3. Chi-Square → Test the relationship between categorical variables.
4. ANOVA → Compare 3 or more means.
5. Mann-Whitney U Test → Non-parametric alternative to the Independent T-Test.
6. Shapiro-Wilk Test → Check whether data is normally distributed before choosing a statistical test.
