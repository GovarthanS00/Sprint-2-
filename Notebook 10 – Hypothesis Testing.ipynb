## Hypothesis Testing
### Definition

Hypothesis Testing is a statistical method used to determine whether there is enough evidence in sample data to support or reject a claim about a population.

It helps us decide whether an observed result happened due to chance or because there is a real effect.

### Steps in Hypothesis Testing
* State the hypotheses.
* Choose a significance level (α).
* Collect sample data.
* Calculate the test statistic.
* Calculate the p-value.
* Compare the p-value with α.
* Make a decision.

```python
### Python Example
print("Hypothesis Testing follows a series of statistical steps.")
```


### Business Example

An online store wants to know whether a new website design increases sales.

### AI/ML Example

A data scientist tests whether a new machine learning model performs significantly better than the current model.

### Key Points
* Uses sample data to make decisions.
* Helps determine whether results are statistically significant.
* Widely used in research, business, and AI/ML.

## 1. Null Hypothesis (H₀)
### Definition

The Null Hypothesis (H₀) assumes that there is no effect, no difference, or no relationship between variables.

It is the default assumption that we try to test.

### Python Example
H0 = "The new model has the same accuracy as the old model."
print(H0)
### Business Example

The new advertisement does not increase sales.

### AI/ML Example

The new machine learning model performs the same as the existing model.

### Key Points
* Default assumption.
* Assumed true until evidence suggests otherwise.
* Usually represents "no change."

## 2. Alternative Hypothesis (H₁)
### Definition

The Alternative Hypothesis (H₁) states that there is an effect, difference, or relationship.

It is accepted only if enough evidence exists against the null hypothesis.

### Python Example
H1 = "The new model performs better than the old model."
print(H1)
### Business Example

The new advertisement increases sales.

### AI/ML Example

The new machine learning model has higher accuracy.

### Key Points
* Opposite of the null hypothesis.
* Accepted when H₀ is rejected.
* Represents the claim being tested.

## 3. p-value
### Definition

The p-value is the probability of obtaining the observed result (or something more extreme) assuming the null hypothesis is true.

A smaller p-value indicates stronger evidence against the null hypothesis.

### Decision Rule
* p-value ≤ α → Reject H₀
* p-value > α → Fail to reject H₀

```python
from scipy.stats import ttest_ind

group1 = [20, 22, 21, 19, 20]
group2 = [25, 24, 23, 26, 24]

t_stat, p_value = ttest_ind(group1, group2)

print("p-value:", p_value)
```

### Business Example

A p-value of 0.02 suggests the new marketing strategy significantly increased sales.

### AI/ML Example

Compare the accuracy of two machine learning models.

### Key Points
Measures evidence against H₀.
Smaller p-value means stronger evidence.
Does not measure the size of the effect.

## 4. Confidence Interval
### Definition

A Confidence Interval (CI) is a range of values that is likely to contain the true population parameter with a certain confidence level (such as 95%)
.

```python
import scipy.stats as stats
import numpy as np

data = [20, 21, 22, 19, 23]

ci = stats.t.interval(
    confidence=0.95,
    df=len(data)-1,
    loc=np.mean(data),
    scale=stats.sem(data)
)

print(ci)
```

### Business Example

The average customer spending is estimated to be between two values with 95% confidence.

### AI/ML Example

Estimate the range of the true accuracy of a classification model.

### Key Points
Gives a range of possible values.
Higher confidence gives a wider interval.
Shows the uncertainty in an estimate.

## 5. Significance Level (α)
### Definition

The Significance Level (α) is the maximum probability of making a Type I Error that we are willing to accept.

The most common value is 0.05.

### Decision Rule
* p-value ≤ α → Reject H₀
* p-value > α → Fail to reject H₀

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

### Business Example

A company accepts a 5% risk of making a wrong decision.

### AI/ML Example

Use α = 0.05 when comparing two machine learning models.

### Key Points
* Common values are 0.05 and 0.01.
* Chosen before the test begins.
* Controls the probability of a Type I Error.

## 6. Type I Error
### Definition

A Type I Error occurs when we reject the null hypothesis even though it is actually true.

It is also called a False Positive.

### Python Example
print("Rejected H0 even though H0 is true.")
### Business Example

Approving a medicine that actually does not work.

### AI/ML Example

Concluding that a new model is better when it actually is not.

### Key Points
* False Positive.
* Probability equals α.
* Controlled by the significance level.

## 7. Type II Error
### Definition

A Type II Error occurs when we fail to reject the null hypothesis even though it is false.

It is also called a False Negative.

### Python Example
print("Failed to reject H0 even though H0 is false.")
### Business Example

Rejecting a marketing strategy that actually improves sales.

### AI/ML Example

Failing to identify that a new model performs better
### Key Points
* False Negative.
* Can be reduced by increasing sample size.
* Lower Type II Error means higher test power.

## 8. One-tailed Test
### Definition

A One-tailed Test checks whether a parameter is greater than or less than a specific value.

The direction is decided before the test.

### Python Example
print("Testing whether the new model performs better.")
### Business Example

Check whether a new machine increases production speed.

### AI/ML Example

Test whether a new model has higher accuracy than the current model.
### Key Points
* Tests only one direction.
* More powerful for directional hypotheses.
* Cannot detect changes in the opposite direction.

## 9. Two-tailed Test
### Definition

A Two-tailed Test checks whether a parameter is different from a specific value, regardless of direction.

### Python Example
print("Testing whether the average value has changed.")
### Business Example

Determine whether customer spending has changed after a pricing update.

### AI/ML Example

Check whether the new model's accuracy is different from the existing model.
### Key Points
* Tests both directions.
* Detects increases and decreases.
* Most commonly used in statistical analysis.
